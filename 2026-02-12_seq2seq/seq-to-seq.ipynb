## Seq2Seq

Limitações da RNN  
- Sequencial, não paralelizada
- Entradas variadas, saídas de tamanho fixas


Seq2Seq é mais utilizado para  
- tradução
- sumarização  
 | não temos uma relação direta entre o tamanho da entrada e de saída  

Seq2Seq == encoder-decoder


encoder -> context -> decoder

Encoder-Decoder com RNNs
- para modelo de tradução, precisamos adicionar um token de fim da saída
- pq? pq senão ele vai ficar gerando infinitamente. O importante não é ele saber o tamanho da saída mas sim a representação das palavras na outra língua

como a saída é gerada?
- passamos o contexto para o decoder
- vai gerar um texto da forma  
the | green | witch | arrived ->  
\<s> | llegó | llegó -> la | la -> bruja | bruja -> verde | verde -> <\/s>


Com textos muito grandes ele não performa tão bem, pois só passamos o contexto no começo do decoder.
- podemos passar o contexto em todos os momentos do decoder, melhora um pouco, mas fica mais pesado

Teacher Forcing

- se erramos uma palavra no início, erramos todo o resto
- vamos ajustando o erro, mostrando qual a palavra certa para proxima predição. O erro aumenta mais suavemente
- podemos usar mais no início do que no fim, ja que no início o erro é maior
