# Transfer Learning

treinar modelos é muito custoso

- gpt 3 demoraria 355 anos pra treinar em 1 só gpu

podemos usar modelos pré treinados atraves do \**fine-tuning**
- pegar uma rede treinada genérica, separando seus dados de treinamento pra um prob. específico
- não precisamos treinar do zero, e precisamos de menos dados
- "especialização" da rede

- usar word embeddings é uma forma de transfer learning
- usualmente, **não retreinamos a rede neural inteira**

- vantagem: reduz a necessidade de dados anotados
- ja começamos com um erro mais baixo

- mantemos alguns pesos/camadas "congelados"
- apenas algumas camadas são treinadas com novos dados, geralmente as últimas
    - não faz sentido treinar as iniciais

- podemos manter o modelo sem mudanças  
e adicionar novas camadas após a última camada do modelo treinado
    - apenas as novas são treinadas com novos dados
- serve muito bem para classificação

- o ideal é não atualizar os pesos existentes, não queremos que o modelo desaprenda
- usar taxa de aprendizagem ajuda a não desfazer o aprendizado anterior
