## LSTM

- tipo de RNN
- trabalha melhor com sequências grandes
- lembrar de informações por longos períodos

bem mais pesado do que rnn

LSTM tem sempre de 4 em 4 camadas, equivalente a 1 camada na rnn
- geralmente chamamos de bloco lstm, ao invés de camada

na LSTM, temos 2 saídas  

X_t + Y_t-1 -> [] -> Y_t  
C_t-1 -> [] -> C_t  

*obs.: [] é um blocao azul
