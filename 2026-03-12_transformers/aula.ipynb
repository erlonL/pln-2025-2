self-attention: tratar melhor contextos grandes
- para cada camada da rede, passamos todos os $x_i$ disponíveis
- dessa forma, a diferença entre RNN é que não esperamos a saída de uma camada para a outra  
- conseguimos paralelizar, computando as saídas ao mesmo tempo

no self-attention, comparamos um vetor com uma **coleção de outros vetores**  
- relação da entrada com a própria entrada
- ex.: talvez tenha um artigo que precisa concordar com um sujeito feminino
- usamos produto escalar para comparação $score(x_i, x_j) = x_ix_j$

dado que o score foi calculado, a ideia do self-attention é a mesma do mecanismo de atenção
- normalização através de softmax, resultando em vetor $\alpha$
- vetor resultante é gerado da soma ponderada dos vetores de entrada
$$y_i = \sum_{}\alpha_{ij}x_j$$

## Query, key e value

Nossa entrada é um vetor  
O embedding vai transformar em 3 entradas diferentes  
Query: foco atual da atenção quando comparado com as entradas anteriores  
- entrada naquele momento que estamos focando (ex.: x_3)  
Key: como uma entrada anterior que é comparada com o foco atual da atenção  
- versão da entrada que vamos comparar a entrada com esse x_3  
- pego uma query e comparo com vários keys
- a key do x_1 e a query do x_1 são diferentes
Value: como valor usado para computar a saída para o foco atual da atenção  


Nota: damos pra arquitetura mais informações daquele dado  

Para cada um dos papéis, temos uma matriz de pesos
- Query: $W^Q$
- Key: $W^K$
- Value: $W^V$
  
em que calculamos seus valores:
- $q_i = W^Qx_i$
- $k_i = W^Kx_i$
- $v_i = W^Vx_i$

- As entradas, saídas e vetores nas camadas intermediárias nos Transformers, todos tem o tamanho $d$
- Vamos assumir por enquanto que $W^Q, W^K, W^V$ todos possuem tamanho $d \times d$

$X$ -> $(d, 1)$  
$W$ -> $(d, d)$  
$q = W.x$ -> $(d, 1)$

## Positional Embeddings

- como modelar posição de cada token no texto?
- Em RNNs a ordem faz parte do funcionamento da rede  

Os Transformers usam a técnica chamada de positional embeddings
- a maneira mais simples é ter outro embedding que representa a posição

mas não é isso que está no paper...
- uma alternativa é utilizar funções ao invés de treinar um novo embedding
- no artigo é utilizada uma função que mistura seno e cosseno


função

Cada dimensão do vetor vai ser calculada por uma função que tem como parâmetros
- posição do token $(pos)$
- Índice da dimensão do vetor que está sendo calculada $(i)$
- O tamanho do vetor $(d_e)$


exemplo  
  
  

$\#1 [0, 0, 0, ...] d$  
$\#2 [0, 0, 0, ...] d$  
$\#3 [0, 0, 0, ...] d$  
  
  
funções

$PE_{pos, 2i} = sin(pos/10000^{frac 2i/d_e})$

e são diferentes para indices pares ou impares, uma usa seno a outra cosseno
