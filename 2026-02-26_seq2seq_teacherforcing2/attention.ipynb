## Atenção

- o encoder constrói uma representaao da entrada (context)
- o decoder usa essa representação para gerar a saída
- o contexto é um gargalo, pois ele precisa representar tudo sobre a entra
- informações sobre o início da entrada podem não ser bem representadas
    - principalmente para sequências
- "tom is waiting for maria" - não precisamos ler "tom" para traduzir "waiting", não tem muito a ver

- o mecanismo de atenção é uma solução para esse gargalo
- o decoder pode pegar informações dos hidden states do decoder
- não conseguimos pegar todos os hidden states sempre, pois a entrada varia
- com o attention criamos um vetor único de tamanho fixo para cada etapa do encoder para ser o contexto
    - esse vetor é uma soma ponderada de todos os hidden states
    - o peso determina em que hidden states o decoder deve focar (prestar atenção)

ex.: temos 3 vetores e somamos esses vetores, cada um podendo ser multiplicado por um escalar (M, N, ou 0)  

M $\times$ [0, 0.12, 0.4, ..., 1]  
N $\times$ [0, 0.12, 0.4, ..., 1]  
0 $\times$ [0, 0.12, 0.4, ..., 1]  

= [0 + 0 + 0, ... 1 + 1 + 1]

assim conseguimos decidir vetores que são mais importantes

como decidimos quais vetores são importantes?

1. Calcular relevância de um hidden state para um determinado passo do encoder
- fazer isso através o produto escalar
- fazer escalar entre o hidden state do decoder com cada um hidden state do encoder
    - isso vai dar um score de relevância de cada hs do encoder em relação ao hs do decoder


- os scores vão gerar um vetor que será normalizado por uma função softmax
- o softmax recebe todos os scores
- pega todos os vetores e gera um valor de no máximo 1
    - o tamanho do vetor resultante do softmax é igual ao tamanho do vetor de entrada

- cada hs do encoder é multiplicado por um resultado do softmax
- [...]

- por fim os vetores são somados gerando apenas um vetor independentemente do valor (tamanho) de entrada
