```python
import torch
from torch.utils.data import Dataset

class IMDBDataset(Dataset):
    def __init__(self, dataset):
        self.data = dataset['tokens_id'] # list[list[int]]
        self.labels = dataset['label']   # list[int] 0/1
    
    def __len__(self):
        return len(self.labels)
    
    def __getitem__(self, index):
        x = torch.tensor(self.data[index], dtype=torch.long)
        y = torch.tensor(self.labels[index], dtype=torch.float32)
        return x, y
```

```python
from torch.utils.data import DataLoader

def collate_batch(batch, pad_id=0):
    xs, ys = zip(*batch)  # tuple of tensors
    lengths = torch.tensor([x.size(0) for x in xs], dtype=torch.long)
    max_len = lengths.max().item()

    x_padded = torch.full((len(xs), max_len), pad_id, dtype=torch.long)
    for i, x in enumerate(xs):
        x_padded[i, :x.size(0)] = x

    # mask: 1 for real tokens, 0 for padding
    mask = (x_padded != pad_id).float()

    y = torch.stack(ys)  # [B]
    return x_padded, mask, y


dataloader = DataLoader(my_dataset, shuffle=True, batch_size=2048, collate_fn=collate_batch)
# batch size: defaulta pra 1, o que faz com que demore bastante
# collate_fn: pode ser um problema ter reviews de diferentes tamanhos, então precisamos de uma função de padding
```

```python
import torch
from torch import nn

class IMDBBinaryClassifier(nn.Module):
    def __init__(self, vocab_size, embedding_dim, pad_id=0):
        super().__init__()
        self.pad_id = pad_id
        self.embed = nn.Embedding(vocab_size, embedding_dim, padding_idx=pad_id)
        self.classifier = nn.Linear(embedding_dim, 1)

    def forward(self, x, mask=None):
        emb = self.embed(x) # [B, T, D]
        
        if mask is None:
            pooled = emb.mean(dim=1) # [B, D]
        else:
            # mean over real tokens only
            mask = mask.unsqueeze(-1)               # [B, T, 1]
            summed = (emb * mask).sum(dim=1)        # [B, D]
            counts = mask.sum(dim=1).clamp(min=1.0) # [B, 1]
            pooled = summed / counts                # [B, D]

        logits = self.classifier(pooled).squeeze(1)
        return logits

```

```python
def train_one_epoch(model, dataloader, optimizer, device):
    model.train()
    loss_fn = nn.BCEWithLogitsLoss()
    total_loss = 0.0

    for x, mask, y in dataloader:
        x = x.to(device)
        mask = mask.to(device)
        y = y.to(device)

        optimizer.zero_grad()
        logits = model(x, mask)          # [B]
        loss = loss_fn(logits, y)        # y must be float 0/1
        loss.backward()
        optimizer.step()

        total_loss += loss.item() * x.size(0)

    return total_loss / len(dataloader.dataset)
```

```python
@torch.no_grad()
def evaluate(model, dataloader, device):
    model.eval()
    loss_fn = nn.BCEWithLogitsLoss()
    total_loss = 0.0
    correct = 0
    total = 0

    for x, mask, y in dataloader:
        x = x.to(device)
        mask = mask.to(device)
        y = y.to(device)

        logits = model(x, mask)
        loss = loss_fn(logits, y)
        total_loss += loss.item() * x.size(0)

        probs = torch.sigmoid(logits)
        preds = (probs >= 0.5).float()
        correct += (preds == y).sum().item()
        total += y.numel()

    return total_loss / len(dataloader.dataset), correct / total
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model = IMDBBinaryClassifier(
    vocab_size=VOCAB_SIZE,
    embedding_dim=128,
    pad_id=0
).to(device)

optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)

# Train one epoch
avg_loss = train_one_epoch(model, dataloader, optimizer, device)
print("train loss:", avg_loss)

# Single batch forward pass (example)
x, mask, y = next(iter(dataloader))
x, mask = x.to(device), mask.to(device)

logits = model(x, mask)                 # [B]
probs = torch.sigmoid(logits)           # [B] in [0,1]
preds = (probs >= 0.5).long()           # [B] 0/1
print(logits[:5], probs[:5], preds[:5])
```
