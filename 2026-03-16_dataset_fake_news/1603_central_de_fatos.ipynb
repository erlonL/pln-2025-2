```python
import pandas as pd
```

```python
df = pd.read_csv("in/central_de_fatos.csv", sep=";")
print(df.shape)
df.head()
```

```python
df[df['rating'] == "['falso', 'falso', 'falso', 'falso']"].iloc[8].to_dict()
```

```python
# parse string to list
import ast
df['rating'] = df['rating'].apply(ast.literal_eval)
df['rating'].head()
```

```python
from collections import Counter
from unidecode import unidecode

# Achata todas as listas e processa cada item
all_items = []
for items_list in df["rating"]:
    for item in items_list:
        all_items.append(unidecode(item).lower())

counter = Counter(all_items)
counter.most_common(10)
```
